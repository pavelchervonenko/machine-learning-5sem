10. Как определить, есть ли в данном двумерном массиве пустые столбцы?

````python
a = np.array([
    [0, 1, 0],
    [0, 2, 0],
    [0, 3, 0]
])

n_rows, n_cols = a.shape

empty_columns = []

for col in range(n_cols):
    column = a[:, col]      
    if (column == 0).all():
        empty_columns.append(col)


if len(empty_columns) > 0:
    print("Пустые столбцы:", empty_columns)
else:
    print("Пустых столбцов нет")
````

2. Сбалансировано ли соотношение мужчин и женщин в наборе данных?

````python
gender_counts = data["Gender"].value_counts()

male = gender_counts.get("Male", 0)
female = gender_counts.get("Female", 0)
diff = female - male

print("Мужчин в наборе:", male)
print("Женщин в наборе:", female)
print()


if male == female:
    print("Да, соотношение полностью сбалансировано")
elif abs(diff) < (0.05 * (male + female)):
    # если разница менее 5%
    print("Да, выборка практически сбалансирована")
elif female > male:
    print("Нет, женщин заметно больше, соотношение несбалансировано")
else:
    print("Нет, мужчин заметно больше, соотношение несбалансировано")
````

3. Правда ли, что люди с высоким уровнем стресса имеют более высокое кровяное давление, чем те, кто не подвержен стрессу

````python
data["Systolic"] = data["Blood Pressure (s/d)"].str.split("/").str[0].astype(float)

stress_median = data['Stress Levels'].median()

high_stress = data[data["Stress Levels"] > stress_median]
low_stress  = data[data["Stress Levels"] < stress_median]

mean_high = high_stress["Systolic"].mean()
mean_low  = low_stress["Systolic"].mean()

print(f"Медиана уровня стресса: {stress_median:.1f}\n")

print(f"Среднее систолическое давление при стрессе выше медианы: {mean_high:.1f}")
print(f"Среднее систолическое давление при стрессе ниже медианы: {mean_low:.1f}\n")

if mean_high > mean_low:
    print("Да, люди с высоким уровнем стресса имеют более высокое давление")
elif mean_high < mean_low:
    print("Нет. Люди с высоким уровнем стресса НЕ имеют более высокого давления")
else:
    print("Давление одинаковое — зависимости не обнаружено")
````

4. Правда ли, что уровень холестерина у мужчин и женщин отличается?

````python
chol_by_gender = data.groupby("Gender")["Cholesterol Level (mg/dL)"].mean()

male_chol   = chol_by_gender.get("Male", None)
female_chol = chol_by_gender.get("Female", None)

diff = abs(male_chol - female_chol)

print(f"Средний холестерин у мужчин: {male_chol:.1f}")
print(f"Средний холестерин у женщин: {female_chol:.1f}")
print(f"Разница: {abs(diff):.1f}\n")

if diff > 0:
    print("Да. Уровень холестерина у мужчин и женщин отличается.")
else:
    print("Нет. Уровень холестерина у мужчин и женщин не отличается.")
````

5. Правда ли, что с возрастом женщины имеют свойство оценивать свои когнитивные способности хуже, чем в юности (порог юности - 30 лет, женщина в возрасте - 60 лет+)

````python
women = data[data["Gender"] == "Female"]

young_women = women[women["Age (years)"] < 30]
older_women = women[women["Age (years)"] >= 60]

mean_young = young_women["Cognitive Function"].mean()
mean_old = older_women["Cognitive Function"].mean()

diff = mean_young - mean_old

print(f"Средняя когнитивная оценка у молодых женщин (<30): {mean_young:.1f}")
print(f"Средняя когнитивная оценка у женщин (>60): {mean_old:.1f}")
print(f"Разница: {diff:.1f}\n")

if mean_young > mean_old:
    print("Да. Пожилые женщины оценивают свои когнитивные способности ниже, чем молодые")
else:
    print("Нет. Пожилые женщины не оценивают свои когнитивные способности хуже")
````

6. Влияет ли физическая активность на вес в возрасте более 50 лет?

````python
older = data[data["Age (years)"] > 50]

weight_by_activity = older.groupby("Physical Activity Level")["Weight (kg)"].mean()

low = weight_by_activity.get("Low", None)
moderate = weight_by_activity.get("Moderate", None)
high = weight_by_activity.get("High", None)

print("Для людей в возрасте более 50 лет:")
print(f"Средний вес при низкой активности:  {low:.1f} кг")
print(f"Средний вес при средней активности: {moderate:.1f} кг")
print(f"Средний вес при высокой активности: {high:.1f} кг\n")

# здесь я проверяю, отличаются ли средние значения между группами активности
unique_means = weight_by_activity.round(1).unique()

if len(unique_means) > 1:
    print("Да. Физическая активность влияет на вес у людей старше 50 лет.")
else:
    print("Нет. По этим данным физическая активность не влияет на вес у людей старше 50 лет.")


````

7. Сравните минимальный вес мужчин с высоким и низким заработком

8. Правда ли, что у людей с индексом массы тела больше 75го перцентиля встречаются хронические заболевания чаще, чем у людей с индесом массы тела в переделах от 30 до 60 перцентиля?

````python
p30 = data["BMI"].quantile(0.30)
p60 = data["BMI"].quantile(0.60)
p75 = data["BMI"].quantile(0.75)

group_high_bmi = data[data["BMI"] > p75]
group_mid_bmi  = data[(data["BMI"] >= p30) & (data["BMI"] <= p60)]

has_disease_high = group_high_bmi["Chronic Diseases"].notna().mean()
has_disease_mid  = group_mid_bmi["Chronic Diseases"].notna().mean()

print(f"Доля заболеваний при BMI > 75 перцентиля: {has_disease_high:.3f}")
print(f"Доля заболеваний при BMI 30–60 перцентиль: {has_disease_mid:.3f}\n")

if has_disease_high > has_disease_mid:
    print("Да. У людей с BMI выше 75-го перцентиля хронические заболевания встречаются чаще")
else:
    print("Нет. У людей с BMI выше 75-го перцентиля хронические заболевания НЕ встречаются чаще")
````

9. Правда ли, что люди с высоким уровнем дохода имеют больший уровень физической активности, чем люди с низким доходом

````python
activity_map = {
    "Low": 1,
    "Moderate": 2,
    "High": 3
}

data["ActivityScore"] = data["Physical Activity Level"].map(activity_map)

high_income = data[data["Income Level"] == "High"]
low_income  = data[data["Income Level"] == "Low"]

avg_high_income_activity = high_income["ActivityScore"].mean()
avg_low_income_activity  = low_income["ActivityScore"].mean()

print(f"Средний уровень активности (High income): {avg_high_income_activity:.2f}")
print(f"Средний уровень активности (Low income): {avg_low_income_activity:.2f}\n")

if avg_high_income_activity > avg_low_income_activity:
    print("Да. Люди с высоким доходом имеют более высокий уровень физической активности.")
else:
    print("Нет. Люди с высоким доходом НЕ имеют более высокого уровня физической активности.")
````

5. Постройте круговую диаграмму отображающую как ментальное здоровье связано с паттернами сна

````python
statuses = data['Mental Health Status'].unique()

sleep_order = ['Normal', 'Insomnia', 'Excessive']

fig, axes = plt.subplots(len(statuses), 1, figsize=(7, 5 * len(statuses)))

if len(statuses) == 1:
    axes = [axes]

for ax, status in zip(axes, statuses):
    subset = data[data['Mental Health Status'] == status]
    counts = subset['Sleep Patterns'].value_counts().reindex(sleep_order).dropna()

    ax.pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%',
        startangle=90
    )
    ax.set_title(f'Sleep Patterns for "{status}" Mental Health')
    ax.set_ylabel('')

plt.tight_layout()
plt.show()
````
